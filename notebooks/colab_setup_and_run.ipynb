```
from pathlib import Path
import zipfile, shutil
PACKAGE_ZIP = '/content/StockAI_India_ML_Engineer_Package.zip'
DEST = Path('/content/drive/MyDrive/stock_market_ai_india_cloud/stockai_ml_package')
if not Path(PACKAGE_ZIP).exists(): raise FileNotFoundError(f'Upload package ZIP to {PACKAGE_ZIP}')
stage=Path('/content/stockai_stage')
if stage.exists(): shutil.rmtree(stage)
stage.mkdir()
with zipfile.ZipFile(PACKAGE_ZIP) as z: z.extractall(stage)
src=stage/'StockAI_India_ML_Engineer_Package'
if not src.exists():
    candidates=[x for x in stage.iterdir() if x.is_dir()]
    if len(candidates)!=1: raise RuntimeError('Could not identify extracted project folder')
    src=candidates[0]
if DEST.exists(): print('Preserving existing destination:',DEST,'; inspect/backup before replacing it.')
else:
    DEST.parent.mkdir(parents=True,exist_ok=True); shutil.copytree(src,DEST)
print('Project path:',DEST)
```
